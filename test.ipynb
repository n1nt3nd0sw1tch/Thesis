```python
import json, sys, urllib.request, urllib.error
sys.path.insert(0, 'scripts')
import utils

body = {'model': 'openai/gpt-oss-safeguard-20b', 'temperature': 0.0,
        'max_completion_tokens': 4096, 'reasoning_effort': 'low',
        'messages': [{'role': 'user', 'content': 'Reply with the single word OK.'}]}
request = urllib.request.Request(
    'https://api.groq.com/openai/v1/chat/completions', method='POST',
    data=json.dumps(body).encode(),
    headers={'Content-Type': 'application/json',
             'Authorization': f"Bearer {utils.api_key('groq')}"})
try:
    with urllib.request.urlopen(request, timeout=60) as response:
        print(json.dumps(json.loads(response.read()), indent=2)[:600])
except urllib.error.HTTPError as problem:
    print(problem.code, problem.read().decode()[:400])
```
